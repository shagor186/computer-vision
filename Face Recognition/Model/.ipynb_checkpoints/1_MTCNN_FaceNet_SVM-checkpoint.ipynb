```python
import os
import cv2 as cv
import numpy as np
from mtcnn.mtcnn import MTCNN
from PIL import Image
from keras_facenet import FaceNet
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
import warnings
warnings.filterwarnings("ignore")
```

```python
# Config
TARGET_SIZE = (160, 160)  # FaceNet default input
ALLOWED_EXT = ('.jpg', '.jpeg', '.png', '.webp', '.bmp')
SAVED_MODEL_PATH = 'svm_face_model.pkl'
EMBEDDING_BATCH_SAVE = None  # optional: save embeddings
```

```python
# FACELOADING: reading, cleaning, converting & face-cropping
class FACELOADING:
    def __init__(self, directory, target_size=TARGET_SIZE):
        self.directory = directory
        self.target_size = target_size
        self.detector = MTCNN()
        self.X = []
        self.Y = []

    def read_and_convert(self, filepath):
        try:
            img = Image.open(filepath).convert("RGB")
        except Exception as e:
            print(f"❌ Bad image removed: {filepath} ({e})")
            try: os.remove(filepath)
            except: pass
            return None, None

        ext = os.path.splitext(filepath)[1].lower()
        if ext not in ('.jpg', '.jpeg'):
            new_path = os.path.splitext(filepath)[0] + ".jpg"
            try:
                img.save(new_path, "JPEG", quality=95)
                if new_path != filepath:
                    try: os.remove(filepath)
                    except: pass
                filepath = new_path
            except Exception as e:
                print(f"❌ Conversion failed: {filepath} ({e})")
                return None, None

        img_np = np.array(img)
        img_bgr = cv.cvtColor(img_np, cv.COLOR_RGB2BGR)
        return img_bgr, filepath

    def extract_face(self, filepath):
        img_bgr, final_path = self.read_and_convert(filepath)
        if img_bgr is None: return None

        img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

        try:
            faces = self.detector.detect_faces(img_rgb)
        except Exception as e:
            print(f"❌ MTCNN error for {final_path}: {e}")
            return None

        if not faces:
            print(f"⚠️ No face found: {final_path}")
            return None

        x, y, w, h = faces[0]['box']
        x, y = abs(x), abs(y)
        h_img, w_img = img_rgb.shape[:2]
        x2, y2 = min(x + w, w_img), min(y + h, h_img)
        face = img_rgb[y:y2, x:x2]

        if face.size == 0 or face.shape[0] < 10 or face.shape[1] < 10:
            print(f"⚠️ Tiny/invalid face crop: {final_path}")
            return None

        face_resized = cv.resize(face, self.target_size)
        return face_resized

    def load_faces_from_class(self, class_dir):
        faces = []
        for fname in os.listdir(class_dir):
            fullpath = os.path.join(class_dir, fname)
            if not os.path.isfile(fullpath):
                continue
            face = self.extract_face(fullpath)
            if face is not None:
                faces.append(face)
        return faces

    def load_classes(self):
        for sub in sorted(os.listdir(self.directory)):
            subpath = os.path.join(self.directory, sub)
            if not os.path.isdir(subpath):
                continue
            faces = self.load_faces_from_class(subpath)
            labels = [sub] * len(faces)
            print(f"✅ Loaded {len(faces)} valid faces for class '{sub}'")
            self.X.extend(faces)
            self.Y.extend(labels)

        X_arr = np.asarray(self.X) if self.X else np.empty((0, *self.target_size, 3))
        Y_arr = np.asarray(self.Y) if self.Y else np.array([])
        return X_arr, Y_arr

    def plot_samples(self, limit=12):
        import math
        if len(self.X) == 0:
            print("No images to show.")
            return
        n = min(limit, len(self.X))
        cols = 4
        rows = math.ceil(n / cols)
        plt.figure(figsize=(cols * 3, rows * 3))
        for i in range(n):
            plt.subplot(rows, cols, i + 1)
            plt.imshow(self.X[i])
            plt.axis('off')
        plt.show()
```

```python
# Embedding using FaceNet
embedder = FaceNet()  # load pretrained FaceNet

def get_embedding(face_img):
    img = face_img.astype('float32')
    img = np.expand_dims(img, axis=0)
    emb = embedder.embeddings(img)
    return emb[0]
```

```python
# Train SVM with metrics
def train_svm_with_metrics(EMBEDDED_X, Y_labels, test_size=0.25, random_state=17):
    encoder = LabelEncoder()
    Y_enc = encoder.fit_transform(Y_labels)

    X_train, X_test, Y_train, Y_test = train_test_split(
        EMBEDDED_X, Y_enc, test_size=test_size, random_state=random_state, stratify=Y_enc
    )

    model = SVC(kernel='linear', probability=True, random_state=random_state)
    model.fit(X_train, Y_train)

    # Predictions
    Y_train_pred = model.predict(X_train)
    Y_test_pred = model.predict(X_test)

    # Metrics
    print("\n🔹 Training Results:")
    print(f"Train Accuracy : {accuracy_score(Y_train, Y_train_pred):.4f}")
    print(f"Test Accuracy  : {accuracy_score(Y_test, Y_test_pred):.4f}")
    print("\n📄 Classification Report:")
    print(classification_report(Y_test, Y_test_pred, target_names=encoder.classes_))

    # Confusion matrix
    cm = confusion_matrix(Y_test, Y_test_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=encoder.classes_, yticklabels=encoder.classes_)
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

    return model, encoder
```

```python
# Predict single image
def predict_image_from_path(model, encoder, image_path):
    detector = MTCNN()
    try:
        img = Image.open(image_path).convert("RGB")
    except Exception as e:
        print(f"Cannot open image: {e}")
        return None

    img_np = np.array(img)
    faces = detector.detect_faces(img_np)
    if not faces:
        print("No face detected in test image.")
        return None

    x, y, w, h = faces[0]['box']
    x, y = abs(x), abs(y)
    h_img, w_img = img_np.shape[:2]
    x2 = min(x + w, w_img)
    y2 = min(y + h, h_img)
    face = img_np[y:y2, x:x2]
    face_resized = cv.resize(face, TARGET_SIZE)

    emb = get_embedding(face_resized).reshape(1, -1)
    pred = model.predict(emb)
    name = encoder.inverse_transform(pred)[0]
    prob = np.max(model.predict_proba(emb))
    print(f"Predicted: {name}  (prob: {prob:.3f})")
    return name, prob
```

```python
# Save / Load Model
def save_model_and_encoder(model, encoder, path=SAVED_MODEL_PATH):
    with open(path, 'wb') as f:
        pickle.dump({'model': model, 'encoder': encoder}, f)
    print(f"Model & encoder saved to {path}")

def load_model_and_encoder(path=SAVED_MODEL_PATH):
    with open(path, 'rb') as f:
        data = pickle.load(f)
    return data['model'], data['encoder']
```

```python
# Full pipeline
def run_pipeline(train_dir, test_image_path=None, plot_samples=True):
    loader = FACELOADING(train_dir, target_size=TARGET_SIZE)
    X_images, Y_labels = loader.load_classes()
    if X_images.size == 0:
        print("No valid faces found in training directory.")
        return

    if plot_samples:
        loader.plot_samples(limit=12)

    print("⚙️ Generating embeddings (this may take a while)...")
    EMBEDDED = np.array([get_embedding(img) for img in X_images])

    if EMBEDDING_BATCH_SAVE:
        np.save(EMBEDDING_BATCH_SAVE, EMBEDDED)
        print(f"Saved embeddings to {EMBEDDING_BATCH_SAVE}.npy")

    model, encoder = train_svm_with_metrics(EMBEDDED, Y_labels)
    save_model_and_encoder(model, encoder)

    if test_image_path:
        print("\n🔎 Predicting test image...")
        predict_image_from_path(model, encoder, test_image_path)

    return model, encoder
```

```python
# Run example
if __name__ == "__main__":
    train_dir = "dataset/train"  # Subfolders: train/personA, train/personB
    test_image = "dataset/test/ronaldo/ronaldo.jpg"  # Optional single test image

    run_pipeline(train_dir, test_image_path=test_image, plot_samples=True)
```

```python
import cv2
import numpy as np
from mtcnn.mtcnn import MTCNN
from keras_facenet import FaceNet
import pickle
from PIL import Image

# ---- Load Saved Model ----
with open("svm_face_model.pkl", "rb") as f:
    data = pickle.load(f)
model = data["model"]
encoder = data["encoder"]

# ---- Load FaceNet ----
embedder = FaceNet()

detector = MTCNN()
TARGET_SIZE = (160, 160)

def get_embedding(face_img):
    img = face_img.astype("float32")
    img = np.expand_dims(img, axis=0)
    emb = embedder.embeddings(img)
    return emb[0]

# ---- Webcam ----
cap = cv2.VideoCapture(0)

print("🎥 Webcam started. Press 'q' to quit.")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    faces = detector.detect_faces(rgb)

    for f in faces:
        x, y, w, h = f['box']
        x, y = abs(x), abs(y)
        x2, y2 = x + w, y + h

        face = rgb[y:y2, x:x2]

        if face.size == 0:
            continue

        try:
            face_resized = cv2.resize(face, TARGET_SIZE)
        except:
            continue

        emb = get_embedding(face_resized).reshape(1, -1)
        pred = model.predict(emb)
        name = encoder.inverse_transform(pred)[0]
        prob = np.max(model.predict_proba(emb))

        # Draw box + name
        cv2.rectangle(frame, (x, y), (x2, y2), (0, 255, 0), 2)
        cv2.putText(frame, f"{name} ({prob:.2f})",
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 0),
                    2)

    cv2.imshow("Real-Time Face Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from mtcnn.mtcnn import MTCNN
from keras_facenet import FaceNet
import pickle
from PIL import Image

# ---- Load Saved Model ----
with open("svm_face_model.pkl", "rb") as f:
    data = pickle.load(f)
model = data["model"]
encoder = data["encoder"]

# ---- Load FaceNet ----
embedder = FaceNet()

detector = MTCNN()
TARGET_SIZE = (160, 160)

def get_embedding(face_img):
    img = face_img.astype("float32")
    img = np.expand_dims(img, axis=0)
    emb = embedder.embeddings(img)
    return emb[0]

# ---- Load Video File ----
video_path = "vid/s.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("❌ Cannot open video file")
    exit()

print("🎬 Video loaded. Press 'q' to quit.")

while True:
    ret, frame = cap.read()
    if not ret:
        print("Video finished.")
        break

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    faces = detector.detect_faces(rgb)

    for f in faces:
        x, y, w, h = f['box']
        x, y = abs(x), abs(y)
        x2, y2 = x + w, y + h

        face = rgb[y:y2, x:x2]
        if face.size == 0:
            continue

        try:
            face_resized = cv2.resize(face, TARGET_SIZE)
        except:
            continue

        emb = get_embedding(face_resized).reshape(1, -1)
        pred = model.predict(emb)
        name = encoder.inverse_transform(pred)[0]
        prob = np.max(model.predict_proba(emb))

        cv2.rectangle(frame, (x, y), (x2, y2), (0, 255, 0), 2)
        cv2.putText(frame, f"{name} ({prob:.2f})",
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 0),
                    2)

    # -----------------------------
    # ⭐ Resize video window (SMALL)
    # -----------------------------
    frame = cv2.resize(frame, (840, 760))   # width=640px, height=360px

    cv2.imshow("Video Face Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
