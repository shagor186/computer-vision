```python
from deepface import DeepFace
import os
import numpy as np
from numpy.linalg import norm

# -----------------------------
# 1️⃣ Dataset path
# -----------------------------
train_folder = "/content/drive/MyDrive/Face Recognition/CNN Face Recognition/img_dataset/train"
test_img_path = "/content/drive/MyDrive/Face Recognition/CNN Face Recognition/img_dataset/test/sinner/15.jpg"

# -----------------------------
# 2️⃣ Build embeddings for train dataset
# -----------------------------
print("Building embeddings for train dataset...")
embeddings_db = []
labels_db = []

for person_name in os.listdir(train_folder):
    person_path = os.path.join(train_folder, person_name)
    if not os.path.isdir(person_path):
        continue
    for img_name in os.listdir(person_path):
        img_path = os.path.join(person_path, img_name)
        try:
            embedding = DeepFace.represent(
                img_path,
                model_name='VGG-Face',
                detector_backend='opencv',
                enforce_detection=False  # face detect miss হলে skip করবে
            )
            embeddings_db.append(embedding[0]["embedding"])
            labels_db.append(person_name)
        except Exception as e:
            print(f"Skipping {img_path}: {e}")

print(f"Total embeddings collected: {len(embeddings_db)}")

# -----------------------------
# 3️⃣ Generate embedding for test image
# -----------------------------
print("Processing test image...")
test_embedding = DeepFace.represent(
    test_img_path,
    model_name='VGG-Face',
    detector_backend='opencv',
    enforce_detection=False
)[0]["embedding"]

# -----------------------------
# 4️⃣ Compute similarity with train embeddings
# -----------------------------
def cosine_similarity(a, b):
    return np.dot(a, b)/(norm(a)*norm(b))

best_score = -1
best_label = None

for emb, label in zip(embeddings_db, labels_db):
    score = cosine_similarity(test_embedding, emb)
    if score > best_score:
        best_score = score
        best_label = label

print(f"\nPredicted Name: {best_label}")
print(f"Similarity Score: {best_score:.4f}")

# -----------------------------
# 5️⃣ Optional: Display image with predicted name
# -----------------------------
import matplotlib.pyplot as plt
import cv2

# OpenCV loads image in BGR, Matplotlib expects RGB
img_cv = cv2.imread(test_img_path)
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)

# Overlay predicted name using OpenCV
cv2.putText(img_rgb, best_label, (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

# Display using matplotlib
plt.figure(figsize=(6,6))
plt.imshow(img_rgb)
plt.axis('off')
plt.title(f"Predicted: {best_label}")
plt.show()
```

```python
import os
import numpy as np
from numpy.linalg import norm
import matplotlib.pyplot as plt
import cv2
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.preprocessing import image
from tensorflow.keras.models import Model

# -----------------------------
# 1️⃣ Dataset paths
# -----------------------------
train_folder = "/content/drive/MyDrive/Face Recognition/CNN Face Recognition/img_dataset/train"
test_folder = "/content/drive/MyDrive/Face Recognition/CNN Face Recognition/img_dataset/test"

# -----------------------------
# 2️⃣ Load MobileNetV2 as feature extractor
# -----------------------------
base_model = MobileNetV2(weights='imagenet', include_top=False, pooling='avg')
model = Model(inputs=base_model.input, outputs=base_model.output)  # output: 1280-D embeddings

# -----------------------------
# # 3️⃣ Cosine similarity function
# # -----------------------------
# def cosine_similarity(a, b):
#     return np.dot(a, b) / (norm(a) * norm(b))

# # 2️⃣ Cosine distance (distance version)
# def cosine_distance(a, b):
#     return 1 - np.dot(a, b) / (norm(a) * norm(b))


# 3️⃣ Euclidean distance (L2 distance)
def euclidean_distance(a, b):
    return norm(a - b)

# -----------------------------
# 4️⃣ Function to get embedding from image
# -----------------------------
def get_embedding(img_path, target_size=(224,224)):
    img = image.load_img(img_path, target_size=target_size)
    x = image.img_to_array(img)
    x = np.expand_dims(x, axis=0)
    x = preprocess_input(x)
    embedding = model.predict(x)
    return embedding[0]

# -----------------------------
# 5️⃣ Build embeddings for train dataset
# -----------------------------
embeddings_db = []
labels_db = []

print("Building embeddings for train dataset...")
for person_name in os.listdir(train_folder):
    person_path = os.path.join(train_folder, person_name)
    if not os.path.isdir(person_path):
        continue
    for img_name in os.listdir(person_path):
        img_path = os.path.join(person_path, img_name)
        try:
            emb = get_embedding(img_path)
            embeddings_db.append(emb)
            labels_db.append(person_name)
        except Exception as e:
            print(f"Skipping {img_path}: {e}")

print(f"Total embeddings collected: {len(embeddings_db)}")

# -----------------------------
# 6️⃣ Loop through test dataset
# -----------------------------
for person_name in os.listdir(test_folder):
    person_path = os.path.join(test_folder, person_name)
    if not os.path.isdir(person_path):
        continue

    for test_img_name in os.listdir(person_path):
        test_img_path = os.path.join(person_path, test_img_name)
        if not os.path.isfile(test_img_path):
            continue

        try:
            test_emb = get_embedding(test_img_path)

            # Find best match
            best_score = -1
            best_label = None
            for emb, label in zip(embeddings_db, labels_db):
                score = cosine_similarity(test_emb, emb)
                if score > best_score:
                    best_score = score
                    best_label = label

            print(f"Test Image: {test_img_name} --> Predicted: {best_label}, Score: {best_score:.4f}")

            # Display image
            img_cv = cv2.imread(test_img_path)
            img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
            cv2.putText(img_rgb, best_label, (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)
            plt.figure(figsize=(5,5))
            plt.imshow(img_rgb)
            plt.axis('off')
            plt.title(f"Predicted: {best_label}")
            plt.show()

        except Exception as e:
            print(f"Skipping {test_img_name}: {e}")
```
